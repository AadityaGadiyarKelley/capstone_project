```python
def process_targets(data_length, early_rul = None):
    if early_rul == None:
        return np.arange(data_length-1, -1, -1)
    else:
        early_rul_duration = data_length - early_rul
        if early_rul_duration <= 0:
            return np.arange(data_length-1, -1, -1)
        else:
            return np.append(early_rul*np.ones(shape = (early_rul_duration,)), np.arange(early_rul-1, -1, -1))
```

This function takes in input data and target data and processes them with a specified window length and shift size. The output_data is a numpy array with shape (num_batches, window_length, num_features). If target_data is not None, the function also outputs output_targets, a numpy array with shape (num_batches,) containing the target values for each batch.

```python
def process_input_data_with_targets(input_data, target_data = None, window_length = 1, shift = 1):
    num_batches = np.int(np.floor((len(input_data) - window_length)/shift)) + 1
    num_features = input_data.shape[1]
    output_data = np.repeat(np.nan, repeats = num_batches * window_length * num_features).reshape(num_batches, window_length,
                                                                                                  num_features)
    if target_data is None:
        for batch in range(num_batches):
            output_data[batch,:,:] = input_data[(0+shift*batch):(0+shift*batch+window_length),:]
        return output_data
    else:
        output_targets = np.repeat(np.nan, repeats = num_batches)
        for batch in range(num_batches):
            output_data[batch,:,:] = input_data[(0+shift*batch):(0+shift*batch+window_length),:]
            output_targets[batch] = target_data[(shift*batch + (window_length-1))]
        return output_data, output_targets
```

```python
def process_test_data(test_data_for_an_engine, window_length, shift, num_test_windows = 1):
      
        max_num_test_batches = np.int(np.floor((len(test_data_for_an_engine) - window_length)/shift)) + 1
        if max_num_test_batches < num_test_windows:
            required_len = (max_num_test_batches -1)* shift + window_length
            batched_test_data_for_an_engine = process_input_data_with_targets(test_data_for_an_engine[-required_len:, :],
                                                                              target_data = None,
        window_length= window_length, shift = shift)
            return batched_test_data_for_an_engine, max_num_test_batches
        else:
            required_len = (num_test_windows - 1) * shift + window_length
            batched_test_data_for_an_engine = process_input_data_with_targets(test_data_for_an_engine[-required_len:, :],
                                                                              target_data = None,
                                                                              window_length= window_length, shift = shift)
            return batched_test_data_for_an_engine, num_test_windows
```

```python
import numpy as np
np.int = int
for i in np.arange(1, num_train_machines + 1):
    temp_train_data = train_data[train_data[0] == i].drop(columns = [0]).values
    
    
    if (len(temp_train_data) < window_length):
        print("Train engine {} doesn't have enough data for window_length of {}".format(i, window_length))
        raise AssertionError("Window length is larger than number of data points for some engines. "
                             "Try decreasing window length.")
        
    temp_train_targets = process_targets(data_length = temp_train_data.shape[0], early_rul = early_rul)
    data_for_a_machine, targets_for_a_machine = process_input_data_with_targets(temp_train_data, temp_train_targets, window_length= window_length, shift = shift)
    
    processed_train_data.append(data_for_a_machine)
    processed_train_targets.append(targets_for_a_machine)

processed_train_data = np.concatenate(processed_train_data)
processed_train_targets = np.concatenate(processed_train_targets)
```

```python
for i in np.arange(1, num_test_machines + 1):
    temp_test_data = test_data[test_data[0] == i].drop(columns = [0]).values
    
    # Determine whether it is possible to extract test data with the specified window length.
    if (len(temp_test_data) < window_length):
        print("Test engine {} doesn't have enough data for window_length of {}".format(i, window_length))
        raise AssertionError("Window length is larger than number of data points for some engines. "
                             "Try decreasing window length.")
     # Prepare test data
    test_data_for_an_engine, num_windows = process_test_data(temp_test_data, window_length=window_length, shift = shift,
                                                             num_test_windows = num_test_windows)
    
    processed_test_data.append(test_data_for_an_engine)
    num_test_windows_list.append(num_windows)

processed_test_data = np.concatenate(processed_test_data)
true_rul = true_rul[0].values

# Shuffle training data
index = np.random.permutation(len(processed_train_targets))
processed_train_data, processed_train_targets = processed_train_data[index], processed_train_targets[index]
```

```python
def create_compiled_model_v2():
    model = Sequential([
        layers.LSTM(64, input_shape=(window_length, num_features), return_sequences=True),
        layers.Dropout(0.3),
        layers.LSTM(32),
        layers.Dropout(0.3),
        layers.Dense(32, activation="relu"),
        layers.Dropout(0.2),
        layers.Dense(1)
    ])
    model.compile(loss="mse", optimizer=tf.keras.optimizers.Adam(learning_rate=0.001))
    return model
```

```python
# =============================================================================
# ENGINE-LEVEL CROSS-VALIDATION
# =============================================================================

from sklearn.model_selection import GroupKFold
import numpy as np

print("\n" + "="*70)
print("PERFORMING ENGINE-LEVEL 5-FOLD CROSS-VALIDATION")
print("="*70)

# We need to rebuild the engine groups AND the data WITHOUT shuffling
# =============================================================================

# Step 1: Rebuild the training data WITHOUT shuffling (temporarily)
print("Rebuilding data with engine labels...")

processed_train_data_unshuffled = []
processed_train_targets_unshuffled = []
engine_groups = []

for engine_id in np.arange(1, num_train_machines + 1):
    temp_train_data = train_data[train_data[0] == engine_id].drop(columns=[0]).values
    
    if len(temp_train_data) < window_length:
        print(f"  Skipping engine {engine_id} (not enough data)")
        continue
    
    # Create RUL targets
    temp_train_targets = process_targets(data_length=temp_train_data.shape[0], early_rul=early_rul)
    
    # Create windows
    data_for_a_machine, targets_for_a_machine = process_input_data_with_targets(
        temp_train_data, temp_train_targets, 
        window_length=window_length, shift=shift
    )
    
    num_windows = len(data_for_a_machine)
    
    # Store data and labels
    processed_train_data_unshuffled.append(data_for_a_machine)
    processed_train_targets_unshuffled.append(targets_for_a_machine)
    engine_groups.extend([engine_id] * num_windows)  # Label each window

# Concatenate
processed_train_data_unshuffled = np.concatenate(processed_train_data_unshuffled)
processed_train_targets_unshuffled = np.concatenate(processed_train_targets_unshuffled)
engine_groups = np.array(engine_groups)

print(f"✓ Total windows: {len(processed_train_data_unshuffled)}")
print(f"✓ Total group labels: {len(engine_groups)}")
print(f"✓ Unique engines: {len(np.unique(engine_groups))}")

# Verify
assert len(processed_train_data_unshuffled) == len(engine_groups), "Mismatch!"

# =============================================================================
# Step 2: Perform GroupKFold CV
# =============================================================================

group_kfold = GroupKFold(n_splits=5)
cv_scores = []

for fold, (train_idx, val_idx) in enumerate(group_kfold.split(
    processed_train_data_unshuffled, 
    processed_train_targets_unshuffled, 
    groups=engine_groups
)):
    print(f"\nFold {fold + 1}/5")
    
    X_train_fold = processed_train_data_unshuffled[train_idx]
    X_val_fold = processed_train_data_unshuffled[val_idx]
    y_train_fold = processed_train_targets_unshuffled[train_idx]
    y_val_fold = processed_train_targets_unshuffled[val_idx]
    
    # Show which engines are in each fold
    train_engines = np.unique(engine_groups[train_idx])
    val_engines = np.unique(engine_groups[val_idx])
    print(f"  Train engines: {len(train_engines)} engines | Val engines: {len(val_engines)} engines")
    
    # NOW we can shuffle within each fold (safe since no engine overlap)
    shuffle_idx = np.random.permutation(len(y_train_fold))
    X_train_fold = X_train_fold[shuffle_idx]
    y_train_fold = y_train_fold[shuffle_idx]
    
    # Create fresh model for each fold
    model_cv = create_compiled_model()
    
    # Train quietly
    history_cv = model_cv.fit(
        X_train_fold, y_train_fold,
        validation_data=(X_val_fold, y_val_fold),
        epochs=10, batch_size=128, verbose=0,
        callbacks=[tf.keras.callbacks.LearningRateScheduler(scheduler, verbose=0)]
    )
    
    # Evaluate on validation fold
    val_pred = model_cv.predict(X_val_fold, verbose=0).flatten()
    val_rmse = np.sqrt(mean_squared_error(y_val_fold, val_pred))
    cv_scores.append(val_rmse)
    
    print(f"  Fold {fold + 1} Validation RMSE: {val_rmse:.2f} cycles")

# =============================================================================
# Step 3: Report Results
# =============================================================================

print("\n" + "="*70)
print(f"ENGINE-LEVEL Cross-Validation Results:")
print(f"  Mean RMSE: {np.mean(cv_scores):.2f} cycles")
print(f"  Std Dev:   {np.std(cv_scores):.2f} cycles")
print(f"  Range:     [{np.min(cv_scores):.2f}, {np.max(cv_scores):.2f}]")
print("="*70)

# Compare to your original CV
print(f"\nCOMPARISON:")
print(f"  Your window-shuffled CV: 9.34 cycles (data leakage)")
print(f"  Corrected engine-level CV: {np.mean(cv_scores):.2f} cycles")
print(f"  Your test RMSE: ~14-15 cycles")
print(f"\nThe corrected CV should now match test performance!")

# Note: Your original processed_train_data (shuffled) is still intact
# We only created processed_train_data_unshuffled for CV validation
# Your subsequent training cells will still use the shuffled version
```

```python
# =============================================================================
# STRATEGY 2: DIRECT RUL PREDICTION FROM LLM
# =============================================================================

import json

system_prompt = """You are an expert in turbofan engine diagnostics.

Your task: Predict the Remaining Useful Life (RUL) based ONLY on sensor readings.

Analyze the sensor patterns for signs of:
- Temperature/pressure anomalies
- Multi-sensor correlations indicating failure modes
- Degradation severity (early, moderate, severe)

FD001 context: Single operating condition, High Pressure Compressor (HPC) degradation.

Respond with ONLY valid JSON, no other text."""

# Initialize predictions
y_pred_llm_direct = preds_for_last_example.copy()

print("Generating Direct LLM RUL predictions...\n")

for engine_id, pred, true, sensor_snapshot in anomalies[:5]:
    
    sensor_report = {name: round(val, 3) for name, val in zip(sensor_names, sensor_snapshot)}
    
    user_prompt = f"""Engine {engine_id}:

Sensor readings:
{json.dumps(sensor_report, indent=2)}

Predict Remaining Useful Life based on these sensors.

Output JSON with:
- predicted_rul: number (your RUL estimate in cycles)
- confidence: number between 0 and 1
- reasoning: brief explanation

Example structure only: {{"predicted_rul": 45, "confidence": 0.75, "reasoning": "temp spike indicates..."}}

Now analyze Engine {engine_id} and provide YOUR prediction."""
    
    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ],
            temperature=0.3,
            response_format={"type": "json_object"}
        )
        
        raw_content = response.choices[0].message.content
        
        if not raw_content or raw_content.strip() == "":
            print(f"Engine {engine_id}: Empty response, using LSTM prediction")
            y_pred_llm_direct[engine_id - 1] = pred
            continue
            
        result = json.loads(raw_content)
        llm_rul = float(result['predicted_rul'])
        
        y_pred_llm_direct[engine_id - 1] = llm_rul
        
        print(f"Engine {engine_id}:")
        print(f"  LSTM: {pred:.1f} | LLM Direct: {llm_rul:.1f} | True: {true:.1f}")
        print(f"  LLM Reasoning: {result.get('reasoning', 'N/A')}\n")
        
    except Exception as e:
        print(f"Engine {engine_id}: Error - {e}")
        y_pred_llm_direct[engine_id - 1] = pred

print(f"Processed {min(5, len(anomalies))} anomalies")
```

```python
# =============================================================================
# COMPARISON: LSTM vs DIRECT LLM
# =============================================================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def compute_s_score(rul_true, rul_pred):
    diff = rul_pred - rul_true
    return np.sum(np.where(diff < 0, np.exp(-diff/13)-1, np.exp(diff/10)-1))

# Get predictions
y_true = true_rul
y_pred_lstm = preds_for_last_example
y_pred_llm = y_pred_llm_direct

# Calculate metrics
# LSTM Baseline
rmse_lstm = np.sqrt(mean_squared_error(y_true, y_pred_lstm))
mae_lstm = mean_absolute_error(y_true, y_pred_lstm)
r2_lstm = r2_score(y_true, y_pred_lstm)
s_score_lstm = compute_s_score(y_true, y_pred_lstm)

# Direct LLM
rmse_llm = np.sqrt(mean_squared_error(y_true, y_pred_llm))
mae_llm = mean_absolute_error(y_true, y_pred_llm)
r2_llm = r2_score(y_true, y_pred_llm)
s_score_llm = compute_s_score(y_true, y_pred_llm)

# Print comparison
print("\n" + "="*70)
print("STRATEGY 2: DIRECT RUL PREDICTION COMPARISON")
print("="*70)

print(f"\n{'Metric':<20} {'LSTM Baseline':<20} {'Direct LLM':<20} {'Difference':<15}")
print("-"*70)

print(f"{'RMSE (cycles)':<20} {rmse_lstm:<20.2f} {rmse_llm:<20.2f} {(rmse_llm - rmse_lstm):>+14.2f}")
print(f"{'MAE (cycles)':<20} {mae_lstm:<20.2f} {mae_llm:<20.2f} {(mae_llm - mae_lstm):>+14.2f}")
print(f"{'R² Score':<20} {r2_lstm:<20.4f} {r2_llm:<20.4f} {(r2_llm - r2_lstm):>+14.4f}")
print(f"{'S-Score':<20} {s_score_lstm:<20.2f} {s_score_llm:<20.2f} {(s_score_llm - s_score_lstm):>+14.2f}")

print("\n" + "="*70)

if rmse_llm < rmse_lstm:
    print("🏆 Direct LLM wins!")
else:
    print("📊 LSTM Baseline performs better")

# Visualizations
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.scatter(y_true, y_pred_lstm, alpha=0.6, color='blue', s=30, label='LSTM')
plt.scatter(y_true, y_pred_llm, alpha=0.6, color='orange', s=30, label='Direct LLM')
plt.plot([0, 130], [0, 130], 'r--', lw=2)
plt.xlabel('True RUL')
plt.ylabel('Predicted RUL')
plt.title('LSTM vs Direct LLM')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 3, 2)
errors_lstm = y_pred_lstm - y_true
errors_llm = y_pred_llm - y_true
plt.hist(errors_lstm, bins=20, alpha=0.5, label='LSTM', color='blue')
plt.hist(errors_llm, bins=20, alpha=0.5, label='Direct LLM', color='orange')
plt.xlabel('Error (Predicted - True)')
plt.ylabel('Frequency')
plt.title('Error Distribution')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 3, 3)
plt.scatter(y_pred_lstm, y_pred_llm, alpha=0.6, s=30)
plt.plot([0, 130], [0, 130], 'r--', lw=2)
plt.xlabel('LSTM Prediction')
plt.ylabel('Direct LLM Prediction')
plt.title('Direct Comparison')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```
